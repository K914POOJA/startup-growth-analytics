```python

import pandas as pd

data = {
    "Channel": ["Instagram", "Google Ads", "Email Marketing"],
    "Spend": [5000, 8000, 2000],
    "Leads": [200, 160, 150],
    "Customers": [20, 24, 30]
}

df = pd.DataFrame(data)

print(df)
```

```python

df["Conversion_Rate"] = (
    df["Customers"] / df["Leads"] * 100
).round(2)

df
```

```python
df["Cost_Per_Lead"] = (
    df["Spend"] / df["Leads"]
).round(2)

df
```

```python

df["CAC"] = (
    df["Spend"] / df["Customers"]
).round(2)

df
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.bar(df["Channel"], df["CAC"])

plt.title("Customer Acquisition Cost by Marketing Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("Cost per Customer (₹)")
plt.xticks(rotation=25)
plt.tight_layout()

plt.show()
```

## Business Insights

1. Email Marketing has the lowest Customer Acquisition Cost (CAC) at approximately ₹66.67 per customer.
2. Email Marketing has the highest conversion rate at 20%.
3. Google Ads has the highest CAC at approximately ₹333.33 per customer.
4. SEO has a CAC of approximately ₹166.67 per customer.
5. Based on this sample dataset, Email Marketing appears to be the most cost-effective channel. More data should be collected before making real business decisions.

**Recommendation:** The business should investigate whether it can expand Email Marketing while continuing to monitor lead quality, customer numbers, and marketing costs.
